````python
#applying comprehension
def initialize_inventory(products):
    return {
        product: int(input(f"Enter the quantity for {product}: "))
        for product in products
    }
````

3. Add a new function to calculate the total price of the customer order.

````python
def total_price(customer_orders):
    
    price_list = [int(input(f"Enter the price of {product}: ")) for product in customer_orders]
    
    return sum(price_list) 
````
