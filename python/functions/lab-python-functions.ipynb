1. Define a function named initialize_inventory

```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]

def initialize_inventory(products):
    inventory_dictionary = {}
    
    for product in products:   # estou a percorrer a lista de produtos
        quantity = int(input(f"Enter the quantity for {product}: "))
        inventory_dictionary[product] = quantity
    
    return inventory_dictionary
```

```python
inventory_dictionary = initialize_inventory(products)
print(inventory_dictionary)
```

2. Define a function named get_customer_orders

```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
```

```python
def get_customer_orders():
    customer_orders = set()

    asking_order = "yes"

    while asking_order == "yes":
        asking_order = (input("Enter your order: "))
        customer_orders.add(asking_order)

        asking_order = input("Do you want to enter place another order? (yes/no): ").lower()
        
    return customer_orders
```

```python
get_customer_orders()
```

3. Define a function named update_inventory that takes customer_orders and inventory as parameters

```python
def update_inventory(customer_orders, inventory):

    for order in customer_orders:

        if order not in inventory: 
            print(f"The product doesn't exist")

        else:
            inventory[order] = inventory[order] - 1
    
    return inventory
```

```python
inventory = initialize_inventory(products)
customer_orders = get_customer_orders()
update_inventory(customer_orders, inventory)
```

4. Define a function named calculate_order_statistics that takes customer_orders and products as parameters.

```python
def calculate_order_statistics(customer_orders, products):
    total_products_ordered = len(customer_orders)   # put sum(customer_orders) before but as a set of strings it can not be calculated

    percentage_unique_products = (len(customer_orders) / len(products)) * 100  #(quantidade produto ordered / quantidade total do produto) * 100

    return total_products_ordered, percentage_unique_products
```

5. Define a function named print_order_statistics that takes order_statistics as a parameter.

```python
def print_order_statistics(orders_statistics):

    total_products_ordered, percentage_unique_products = order_statistics  # tuple for example (5, 30) = order_statistics

    print(f"The total of products ordered is: {total_products_ordered}")
    print(f"The percentage of unique products ordered is: {percentage_unique_products}")
```
