## Exercise 2: Cleaning invalid Values

The dataset contains columns with inconsistent and incorrect values that could affect the accuracy of our analysis. Therefore, we need to clean these columns to ensure that they only contain valid data.

Note that this exercise will focus only on cleaning inconsistent values and will not involve handling null values (NaN or None).

*Hint*:
- *Gender column contains various inconsistent values such as "F", "M", "Femal", "Male", "female", which need to be standardized, for example, to "M" and "F".*
- *State abbreviations be can replaced with its full name, for example "AZ": "Arizona", "Cali": "California", "WA": "Washington"*
- *In education, "Bachelors" could be replaced by "Bachelor"*
- *In Customer Lifetime Value, delete the `%` character*
- *In vehicle class, "Sports Car", "Luxury SUV" and "Luxury Car" could be replaced by "Luxury"*

```python
df["gender"].unique()
df["state"].unique()
df["education"].unique()
df["customer_lifetime_value"].unique()
df["vehicle_class"].unique()

print(df["gender"].unique())
print(df["state"].unique())
print(df["education"].unique())
print(df["customer_lifetime_value"].unique())
print(df["vehicle_class"].unique())
```

```python
# cleaning gender
def clean_gender(x):
    x = str(x).lower()
    if x.startswith("f"):
        return "F"
    elif x.startswith("m"):
        return "M"
    return None 

df["gender"] = df["gender"].apply(clean_gender)
df["gender"]
```

```python
df["gender"].unique
```

```python
#cleaning state
df["state"] = df["state"].str.lower()

df["state"] = df["state"].replace({"az": "Arizona", "wa": "Washington", "cali": "California"})
```

```python
df["state"].unique()
```

```python
#cleaning education
def clean_education(x):
    x = str(x)
    if x == "Bachelors":
        return "Bachelor"
    return x

df["education"] = df["education"].apply(clean_education)
df["education"]
```

```python
df["education"].unique()
```

```python
#cleaning customer_lifetime_value
df["customer_lifetime_value"] = df["customer_lifetime_value"].str.replace("%", "")
df["customer_lifetime_value"]
```

```python
df["vehicle_class"] = df["vehicle_class"].replace({
    "Sports Car": "Luxury",
    "Luxury SUV": "Luxury",
    "Luxury Car": "Luxury"
})

df["vehicle_class"].unique()
```

```python
#before cleaning
df.info()
```

```python
#numbers of missing value in each column
df.isna().sum()
```

```python
##Create a working copy
##To ensure the original dataset remains unchanged
##All cleaning steps will be applied to this copy
df2 = df.copy()
```

```python
# lets fill the categorical columns - gender, state, policy type and vehicle class - using the mode (most common value)
df2['customer'] = df2['customer'].fillna(df2['customer'].mode()[0])
df2['gender'] = df2['gender'].fillna(df2['gender'].mode()[0])
df2['state'] = df2['state'].fillna(df2['state'].mode()[0])
df2['policy_type'] = df2['policy_type'].fillna(df2['policy_type'].mode()[0])
df2['vehicle_class'] = df2['vehicle_class'].fillna(df2['vehicle_class'].mode()[0])
```

```python
# lets fill the numerical columns - customer lifetime value, income, monthly premium auto and number of open complaints - using the mean
df2['customer_lifetime_value'] = df2['customer_lifetime_value'].fillna(df2['customer_lifetime_value'].mean())
df2['income'] = df2['income'].fillna(df2['income'].mean())
df2['monthly_premium_auto'] = df2['monthly_premium_auto'].fillna(df2['monthly_premium_auto'].mean())
df2['total_claim_amount'] = df2['total_claim_amount'].fillna(df2['total_claim_amount'].mean())
```

```python
# lets remove the column of number_of_open_complaints bc it has 100% of NaN so it is useless
df2 = df2.drop(columns=['number_of_open_complaints'])
```
