1. Create a new DataFrame that only includes customers who:
   - have a **low total_claim_amount** (e.g., below $1,000),
   - have a response "Yes" to the last marketing campaign.

```python
df.columns = df.columns.str.lower().str.replace(" ", "_")  # first to uniformizing columns' names

low_claim_yes = df[
    (df["total_claim_amount"] < 1000) &
    (df["response"] == "Yes")
]

low_claim_yes.head()
```

2. Using the original Dataframe, analyze:
   - the average `monthly_premium` and/or customer lifetime value by `policy_type` and `gender` for customers who responded "Yes", and
   - compare these insights to `total_claim_amount` patterns, and discuss which segments appear most profitable or low-risk for the company.

```python
#filter who answered "yes"
yes_df = df[df["response"] == "Yes"]
```

```python
#calculating mean by policy_type and gender
agg_yes = yes_df.groupby(["policy_type", "gender"]).agg(
    avg_monthly_premium=("monthly_premium_auto", "mean"),
    avg_clv=("customer_lifetime_value", "mean"),
    avg_total_claim=("total_claim_amount", "mean")
).reset_index()

agg_yes
```

3. Analyze the total number of customers who have policies in each state, and then filter the results to only include states where there are more than 500 customers.

```python
customers_by_state = df.groupby("state").agg(
    num_customers=("customer", "count")
).reset_index()

customers_by_state
```

```python
#  filtering states where more than 500 customers
big_states = customers_by_state[customers_by_state["num_customers"] > 500]
big_states
```

#insights:
those are the states with higher number of customers, and where the company should invest more time and resources, bc there is where the potential for profit is higher

4. Find the maximum, minimum, and median customer lifetime value by education level and gender. Write your conclusions.

```python
clv_stats = df.groupby(["education", "gender"]).agg(
    clv_min=("customer_lifetime_value", "min"),
    clv_max=("customer_lifetime_value", "max"),
    clv_median=("customer_lifetime_value", "median")
).reset_index()

clv_stats
```

#insights:
Customers with High School or Below show the highest median CLV, especially men — this segment has strong long‑term value.

The Bachelor and College groups have stable and predictable CLV, making them low‑risk and easy to target.

Doctor and Master customers show lower maximum CLV, suggesting limited upside potential.

Across almost all education levels, men have slightly higher median CLV than women.

Education level does not drastically change median CLV

The High School or Below (M) segment has the largest CLV range, indicating both high‑value and low‑value customers

5. The marketing team wants to analyze the number of policies sold by state and month. Present the data in a table where the months are arranged as columns and the states are arranged as rows.

```python
df["effective_to_date"] = pd.to_datetime(df["effective_to_date"])
df["month"] = df["effective_to_date"].dt.month_name()  
```

```python
policies_by_state_month = df.groupby(["state", "month"]).agg(
    policies_sold=("policy", "count")
).reset_index()

policies_by_state_month.head()
```

```python
table_state_month = policies_by_state_month.pivot(
    index="state",
    columns="month",
    values="policies_sold"
).fillna(0)

table_state_month
```

6.  Display a new DataFrame that contains the number of policies sold by month, by state, for the top 3 states with the highest number of policies sold.

*Hint:*
- *To accomplish this, you will first need to group the data by state and month, then count the number of policies sold for each group. Afterwards, you will need to sort the data by the count of policies sold in descending order.*
- *Next, you will select the top 3 states with the highest number of policies sold.*
- *Finally, you will create a new DataFrame that contains the number of policies sold by month for each of the top 3 states.*

```python
total_policies_by_state = policies_by_state_month.groupby("state")["policies_sold"].sum().reset_index()

top3_states = total_policies_by_state.sort_values("policies_sold", ascending=False).head(3)
top3_states
```

```python
top3_names = top3_states["state"].tolist()

top3_policies_by_state_month = policies_by_state_month[
    policies_by_state_month["state"].isin(top3_names)
]

top3_table = top3_policies_by_state_month.pivot(
    index="state",
    columns="month",
    values="policies_sold"
).fillna(0)

top3_table
```

7. The marketing team wants to analyze the effect of different marketing channels on the customer response rate.

Hint: You can use melt to unpivot the data and create a table that shows the customer response rate (those who responded "Yes") by marketing channel.

External Resources for Data Filtering: https://towardsdatascience.com/filtering-data-frames-in-pandas-b570b1f834b9

```python
channel_response = df.groupby("sales_channel").agg(
    total_customers=("customer", "count"),
    yes_customers=("response", lambda x: (x == "Yes").sum())
).reset_index()

channel_response["yes_rate"] = channel_response["yes_customers"] / channel_response["total_customers"]

channel_response
```
